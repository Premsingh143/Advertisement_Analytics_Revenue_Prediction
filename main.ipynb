```python
# Q1: Data Quality Issues & Treatment
"""
DATA QUALITY ISSUES IDENTIFIED:

1. Negative Revenue Values:
   - Issue: Revenue_in_USD has negative values (e.g., -404.47 on 2024-08-01, -70.63 on 2023-07-09)
   - Impact: Skews metrics, breaks ROAS calculation
   - Treatment: Investigate (returns/refunds), keep as is or cap at 0

2. Extremely High Revenue Outliers:
   - Issue: 2025-10-25 has $67,867.65 (50x higher than average)
   - Impact: Distorts analysis, may be data error
   - Treatment: Flag as outlier, analyze separately

3. Zero/Extremely Low Media Spend Days:
   - Issue: Some days have very low spend (e.g., $87.69)
   - Impact: ROAS becomes infinite or very high
   - Treatment: Handle in ROAS calculation with min spend threshold

4. Promo Calendar Gaps:
   - Issue: 988 days without promotions
   - Impact: Limited promo data for analysis
   - Treatment: Flag promo vs non-promo for comparison

5. Mixed Date Ranges:
   - Issue: Some channels start at different dates
   - Impact: Incomplete daily data
   - Treatment: Use available data, note limitations
"""

# Ensure merged_df exists before this block
# If merged_df is not defined, create it first
if 'merged_df' not in locals():
    # Create spend_wide and merged_df if not already created
    spend_wide = spend.pivot_table(
        index="Date",
        columns="Channel",
        values="Media Spend_USD",
        aggfunc="sum"
    ).reset_index()
    
    merged_df = orders.merge(spend_wide, on="Date", how="left")
    
    # Calculate Total_Media_Spend if not already done
    if 'Total_Media_Spend' not in merged_df.columns:
        merged_df["Total_Media_Spend"] = (
            merged_df["Email"] +
            merged_df["Google Ads"] +
            merged_df["Meta"] +
            merged_df["YouTube"]
        )

# Check for negative revenue
negative_revenue = merged_df[merged_df['Revenue_in_USD'] < 0]
print(f"Number of days with negative revenue: {len(negative_revenue)}")
print(negative_revenue[['Date', 'Revenue_in_USD']])

# Check for zero media spend
zero_spend = merged_df[merged_df['Total_Media_Spend'] == 0]
print(f"Days with zero total media spend: {len(zero_spend)}")

# Check for outliers using IQR method
Q1 = merged_df['Revenue_in_USD'].quantile(0.25)
Q3 = merged_df['Revenue_in_USD'].quantile(0.75)
IQR = Q3 - Q1
outliers = merged_df[(merged_df['Revenue_in_USD'] < Q1 - 1.5*IQR) | 
                      (merged_df['Revenue_in_USD'] > Q3 + 1.5*IQR)]
print(f"Revenue outliers detected: {len(outliers)} days")
print(f"Max outlier: ${outliers['Revenue_in_USD'].max():,.2f} on {outliers[outliers['Revenue_in_USD'] == outliers['Revenue_in_USD'].max()]['Date'].iloc[0]}")

# Treatment: Create clean versions for analysis
merged_df['Revenue_clean'] = merged_df['Revenue_in_USD'].clip(lower=0)
merged_df['Revenue_trimmed'] = merged_df['Revenue_in_USD'].clip(
    lower=Q1 - 3*IQR, 
    upper=Q3 + 3*IQR
)
merged_df['Is_Outlier'] = (merged_df['Revenue_in_USD'] > Q3 + 3*IQR) | (merged_df['Revenue_in_USD'] < Q1 - 3*IQR)
```

```python
# Insight 4:Lagged Effects Analysis (Ad Spend Delay Impact)

def calculate_lagged_correlation(df, spend_col, revenue_col, max_lag=14):
    """Calculate correlation between spend and revenue at different lags"""
    correlations = []
    for lag in range(max_lag + 1):
        if lag == 0:
            corr = df[spend_col].corr(df[revenue_col])
        else:
            # Shift revenue forward to see if past spend predicts future revenue
            shifted_revenue = df[revenue_col].shift(-lag)
            # Only use valid pairs (not NaN)
            valid_data = df[[spend_col, revenue_col]].copy()
            valid_data['shifted_revenue'] = shifted_revenue
            valid_data = valid_data.dropna()
            if len(valid_data) > 10:  # Ensure enough data points
                corr = valid_data[spend_col].corr(valid_data['shifted_revenue'])
            else:
                corr = np.nan
        correlations.append(corr)
    return correlations

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
channels = ['Google Ads', 'Meta', 'YouTube', 'Email']
colors = ['#4ECDC4', '#45B7D1', '#96CEB4', '#FF6B6B']

for idx, channel in enumerate(channels):
    ax = axes[idx//2, idx%2]
    
    # Calculate lagged correlations
    lags = range(15)
    correlations = calculate_lagged_correlation(merged_df, channel, 'Revenue_in_USD', max_lag=14)
    
    # Plot
    ax.plot(lags, correlations, marker='o', color=colors[idx], linewidth=2, markersize=8)
    ax.axhline(y=0, color='black', linestyle='-', alpha=0.2)
    base_corr = merged_df[channel].corr(merged_df['Revenue_in_USD'])
    ax.axhline(y=base_corr, color='red', linestyle='--', alpha=0.5, label=f'Base r = {base_corr:.3f}')
    
    # Find max correlation lag (excluding lag 0)
    if len(correlations) > 1:
        corr_without_lag0 = correlations[1:]
        if not all(np.isnan(c) for c in corr_without_lag0):
            max_corr_lag = np.nanargmax(corr_without_lag0) + 1
            max_corr_value = correlations[max_corr_lag]
            
            ax.scatter(max_corr_lag, max_corr_value, color='red', s=100, zorder=5)
            ax.annotate(f'Peak: {max_corr_lag} days\nr={max_corr_value:.3f}', 
                        xy=(max_corr_lag, max_corr_value), 
                        xytext=(max_corr_lag + 1, max_corr_value - 0.1),
                        arrowprops=dict(arrowstyle='->', color='red'),
                        fontsize=9)
    
    ax.set_xlabel('Lag (Days)', fontsize=11)
    ax.set_ylabel('Correlation with Revenue', fontsize=11)
    ax.set_title(f'{channel} Spend Lagged Correlation', fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.set_ylim(-0.3, 0.5)

plt.suptitle('Ad Spend Lag Effect on Revenue', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# Print summary of optimal lag times
print("\n" + "="*50)
print("Optimal Lag Times for Ad Spend")
print("="*50)
for channel in channels:
    correlations = calculate_lagged_correlation(merged_df, channel, 'Revenue_in_USD', max_lag=14)
    # Find max correlation lag (excluding lag 0)
    corr_without_lag0 = correlations[1:]
    if not all(np.isnan(c) for c in corr_without_lag0):
        optimal_lag = np.nanargmax(corr_without_lag0) + 1
        max_corr = correlations[optimal_lag]
        print(f"{channel}: Optimal lag = {optimal_lag} days (r = {max_corr:.3f})")
    else:
        print(f"{channel}: No significant lag effect detected")
```

```python
# Add lag and MA features that exist
lag_cols = [f'Revenue_lag_{lag}' for lag in [1, 3, 7, 14, 30] if f'Revenue_lag_{lag}' in model_data.columns]
ma_cols = [f'Revenue_MA_{window}' for window in [7, 14, 30] if f'Revenue_MA_{window}' in model_data.columns]
spend_lag_cols = [f'Total_Spend_lag_{lag}' for lag in [1, 3, 7] if f'Total_Spend_lag_{lag}' in model_data.columns]

feature_cols.extend(lag_cols)
feature_cols.extend(ma_cols)
feature_cols.extend(spend_lag_cols)

X = model_data[feature_cols]
y = model_data['Revenue_in_USD']
```

```python
# Try multiple models
models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, random_state=42)
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    results[name] = {
        'R2': r2_score(y_test, y_pred),
        'MAE': mean_absolute_error(y_test, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_test, y_pred)),
        'Model': model
    }
    print(f"\n{name}:")
    print(f"  R²: {results[name]['R2']:.4f}")
    print(f"  MAE: ${results[name]['MAE']:,.2f}")
    print(f"  RMSE: ${results[name]['RMSE']:,.2f}")
```

```python
# Step 5: Hyperparameter Tuning (Random Forest)
print("\n" + "-"*40)
print("Hyperparameter Tuning for Random Forest")
print("-"*40)

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, None],
    'max_features': ['sqrt', 'log2'],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

rf = RandomForestRegressor(random_state=42, n_jobs=-1)
grid_search = GridSearchCV(rf, param_grid, cv=3, scoring='r2', n_jobs=-1, verbose=0)
grid_search.fit(X_train, y_train)

print(f"Best parameters: {grid_search.best_params_}")
print(f"Best CV R²: {grid_search.best_score_:.4f}")

best_model = grid_search.best_estimator_
y_pred_tuned = best_model.predict(X_test)

print(f"\nTuned Random Forest Performance:")
print(f"  R²: {r2_score(y_test, y_pred_tuned):.4f}")
print(f"  MAE: ${mean_absolute_error(y_test, y_pred_tuned):,.2f}")
print(f"  RMSE: ${np.sqrt(mean_squared_error(y_test, y_pred_tuned)):,.2f}")
```
